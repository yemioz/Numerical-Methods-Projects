### Gauss-Jordan Method

Gauss-Jordan Elimination is an algorithm that can be used to solve systems of linear equations. It relies upon three elementary row operations one can use on a matrix:

1.Swap the positions of two of the rows.

2.Multiply one of the rows by a scalar.

3.Add or subtract the scalar multiple of one row to another row.


Below are functions for these operations.

```python
def swap_rows(matrix, row1, row2):
  new_matrix = []
  for i in range(len(matrix)):
    if i != row1:
      if i != row2:
        new_matrix.append(matrix[i,:])
      else:
        new_matrix.append(matrix[row1,:])
    else:
      new_matrix.append(matrix[row2,:])
  return np.array(new_matrix)


def add_rows(matrix, target_row, modifier_row, value):
  new_row = matrix[target_row,:] + value*matrix[modifier_row,:]
  new_matrix = []
  for i in range(len(matrix)):
    if i != target_row:
      new_matrix.append(matrix[i,:])
    else:
      new_matrix.append(new_row)
  return np.array(new_matrix)


def constant_mult(matrix, row, value):
  new_row = value*matrix[row,:]
  new_matrix = []
  for i in range(len(matrix)):
    if i != row:
      new_matrix.append(matrix[i,:])
    else:
      new_matrix.append(new_row)
  return np.array(new_matrix)
```

Our system of equations with three unknowns is:

\begin{align*}
\left\{
\begin{array}{ll}
      -x_1 - 5x_2 - 5x_3 = 2\\
4x_1 - 5x_2 + 4x_3 = 19\\
x_1 + 5x_2 - x_3 = -20
\end{array}
\right.
\end{align*}


We will now use our previously defined functions to perform the Gauss-Jordan process.



```python
matrix = np.array([[-1, -5, -5, 2], [4, -5, 4, 19], [1, 5, -1, -20]])
```

Below is our system of equations in  matrix form.

```python
matrix
```

The first step is $$row_1 \leftrightarrow row_3$$ to get a leading 1.


```python
matrix = swap_rows(matrix, 0, 2)
matrix
```

Our second step is: $$row_2 - 4row_1$$

```python
matrix = add_rows(matrix, 1, 0, -4)
matrix
```

Next:$$row_3 + row_1$$

```python
matrix = add_rows(matrix, 2, 0, 1)
matrix
```

Then:  $$-\frac{1}{25}row_2$$

```python
matrix = constant_mult(matrix, 1, -1/25)
matrix
```

Next: $$-\frac{1}{6}row_3$$

```python
matrix = constant_mult(matrix, 2, -1/6)
matrix
```

Then: $$row_1+row_3$$

```python
matrix = add_rows(matrix, 0, 2, 1)
matrix
```

Then: $$row_2+\frac{8}{25}row_3$$

```python
matrix = add_rows(matrix, 1, 2, 8/25)
matrix
```

Our final step: $$row_1-5row_2$$

```python
matrix = add_rows(matrix, 0, 1, -5)
matrix
```

As you can see, the solution to our system of equations is: $$x_1 = -2, \text{ } x_2 = -3, \text{ } x_3 = 3$$

Now, building our Gauss-Jordan algorithm:

```python
def smallest_pivot(matrix, row):
  element = matrix[row:,row]
  element = np.abs(element)
  spot = np.where(element == min(element))
  return spot[0][0]+row


def gauss_jordan(matrix):
  new_matrix = matrix
  for control in range(len(matrix)):
    spot = smallest_pivot(new_matrix, control)
    new_matrix = swap_rows(new_matrix, spot, control)
    new_matrix = constant_mult(new_matrix, control, 1/control)
    for i in range(0, control):
      new_matrix = add_rows(new_matrix, i, control, -new_matrix[i,control])
    for i in range(control+1, len(matrix)):
      new_matrix = add_rows(new_matrix, i, control, -new_matrix[i,control])
    return np.array(new_matrix)
```

```python
gauss_jordan(matrix)
```

The problem with building our algorithm comes when when we find our pivot element and try to change it to a leading one. During the algorithm, we divide by zero, which breaks the method. We followed Dr. Jacob's example and catered it to our functions to try to correct it.
